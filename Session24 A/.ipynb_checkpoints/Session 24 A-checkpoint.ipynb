```python
import pandas as pd
import sqlite3 as sql
```

```python
conn = sql.connect('factbook.db')
conn
```

```python
def sql_query(q):
    return pd.read_sql_query(q,conn)
```

```python
q = 'select * from sqlite_master'
```

```python
q = """select * from facts
   where name =='Pakistan'

"""
sql_query(q)
```

```python
q = """select * from cities
    where facts_id == 132

"""
sql_query(q)
```

```python
q = """select * from facts f
       inner join cities c
       on
       f.id == c.facts_id
    

"""
sql_query(q)
```

```python
q = """select f.name Country_name, c.name City from facts f
       inner join cities c
       on
       f.id == c.facts_id
       where f.name == "Pakistan" 
    

"""
sql_query(q)
```

```python
q = """select
           f.name AS Country_name,
           c.name AS City
       from facts f
       left join cities c
       On
       f.id = c.facts_id
       where c.id is null
"""

sql_query(q)
```

```python
q = """select f.name AS Country_name,
              c.name AS City
       from cities c
       right join facts f
       on
       f.id = c.facts_id
       where c.id is null
"""

sql_query(q)
```

```python
q = """SELECT 
           f.name AS Country_name,
           c.name AS City,
           c.population AS City_population
       FROM facts f
       INNER JOIN cities c
       ON 
       f.id = c.facts_id
       WHERE c.population > 10000000
"""

sql_query(q)
```

```python
q = """select f.name Country_name, c.name City from facts f
       inner join cities c
       on
       f.id == c.facts_id
       where f.name == "United Arab Emirates" 
    

"""
sql_query(q)
```

```python
q = """select f.name country_name, c.name city_name, c.population  City_population
      
"""

sql_query(q)
```
